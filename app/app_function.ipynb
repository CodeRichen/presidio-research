```python
import sys
from pathlib import Path
from presidio_analyzer import AnalyzerEngine
from presidio_analyzer.nlp_engine import NlpEngineProvider
import time
import shutil
from pathlib import Path

# 1. 建立本地 NLP 引擎設定（指定使用本地的 en_core_web_sm，不走網路下載）
configuration = {
    "nlp_engine_name": "spacy",
    "models": [{"lang_code": "en", "model_name": "en_core_web_sm"}],
}

print("正在離線初始化 Presidio 引擎...")
provider = NlpEngineProvider(nlp_configuration=configuration)
nlp_engine = provider.create_engine()

# 2. 將本地引擎傳給 AnalyzerEngine
analyzer = AnalyzerEngine(nlp_engine=nlp_engine)
print("✅ 離線引擎初始化完成！\n")

found_count = 0
line_number = 0

def scan_file_for_pii(filepath: str, analyzer: AnalyzerEngine):
    global found_count, line_number
    path = Path(filepath)
    if not path.exists():
        print(f"❌ 錯誤：找不到檔案 '{filepath}'")
        return

    print(f"\n==================================================")
    print(f"📄 開始掃描檔案：{filepath}")
    print(f"==================================================\n")



    # 逐行讀取檔案，降低記憶體消耗
    with path.open("r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            line_number += 1
            
            # 調用 Presidio 分析該行文字
            results = analyzer.analyze(text=line, language="en")

            for result in results:
                found_count += 1
                # 擷取被辨識出的敏感詞語文字內容
                sensitive_word = line[result.start:result.end]
                if result.score >= 0.8:  # 可調整信心度閾值
                    print(
                        f"[{found_count}] 第 {line_number} 行 "
                        f"| 類別: {result.entity_type:<15} "
                        f"| 信心度: {result.score:.2f} "
                        f"| 敏感詞語: '{sensitive_word}'"
                    )
    


analyzer = AnalyzerEngine(nlp_engine=nlp_engine)
print("✅ 引擎初始化完成！\n")

start_time = time.perf_counter()
root = r"D:\2.programm2\Exam\commission3"
folder = Path(root)
for file in folder.rglob("*"):
    if file.is_file() and file.suffix.lower() in {".txt", ".ipynb", ".csv", ".json", ".xml", ".yml", ".yaml", ".md", ".py", ".js", ".ts", ".java", ".c", ".cpp", ".h", ".ini", ".conf", ".env", ".sql", ".html", ".htm", ".sh"}:
        scan_file_for_pii(file, analyzer)

end_time = time.perf_counter()
print(f"\n總共耗時: {end_time - start_time:.2f} 秒")


```
